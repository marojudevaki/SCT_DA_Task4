```python
import pandas as pd

df = pd.read_csv('/content/campaign_data.csv')

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
```

```python
# Convert date columns to datetime format
df['Start_Date'] = pd.to_datetime(df['Start_Date'])
df['End_Date'] = pd.to_datetime(df['End_Date'])

print("Date columns converted successfully!")
print(df[['Start_Date', 'End_Date']].dtypes)
```

```python
print("Missing values after date conversion:")
print(df[['Start_Date', 'End_Date']].isnull().sum())
```

```python
print("Final dataset shape:", df.shape)
print("Total duplicate rows:", df.duplicated().sum())
print("Total missing values:", df.isnull().sum().sum())
```

```python
# Basic statistical summary
df.describe()
```

```python
# Marketing channels/platforms
print("Marketing Platforms:")
print(df['Platform'].value_counts())
```

```python
# Campaign types
print("Campaign Types:")
print(df['Campaign_Type'].value_counts())
```

```python
import matplotlib.pyplot as plt

# Calculate average ROI for each marketing platform
roi_by_platform = df.groupby('Platform')['ROI_%'].mean().sort_values(ascending=False)

print("Average ROI by Marketing Platform:")
print(roi_by_platform)
```

```python
plt.figure(figsize=(10, 6))

roi_by_platform.plot(kind='bar')

plt.title('Average ROI by Marketing Platform')
plt.xlabel('Marketing Platform')
plt.ylabel('Average ROI (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Calculate total funnel metrics
funnel = {
    'Impressions': df['Impressions'].sum(),
    'Clicks': df['Clicks'].sum(),
    'Leads': df['Leads_Generated'].sum(),
    'Conversions': df['Conversions'].sum()
}

print("Marketing Funnel:")
for stage, value in funnel.items():
    print(stage, ":", value)
```

```python
# Funnel visualization
plt.figure(figsize=(10, 6))

stages = list(funnel.keys())
values = list(funnel.values())

plt.bar(stages, values)

plt.title('Marketing Campaign Funnel')
plt.xlabel('Funnel Stage')
plt.ylabel('Number of Users')
plt.tight_layout()
plt.show()
```

```python
# Calculate funnel conversion rates

impressions = df['Impressions'].sum()
clicks = df['Clicks'].sum()
leads = df['Leads_Generated'].sum()
conversions = df['Conversions'].sum()

click_rate = (clicks / impressions) * 100
lead_rate = (leads / clicks) * 100
conversion_rate = (conversions / leads) * 100
overall_conversion = (conversions / impressions) * 100

print(f"Click Rate: {click_rate:.2f}%")
print(f"Lead Rate from Clicks: {lead_rate:.2f}%")
print(f"Conversion Rate from Leads: {conversion_rate:.2f}%")
print(f"Overall Conversion Rate: {overall_conversion:.2f}%")
```

```python
# Marketing funnel chart

stages = ['Impressions', 'Clicks', 'Leads', 'Conversions']
values = [impressions, clicks, leads, conversions]

plt.figure(figsize=(10, 6))
plt.bar(stages, values)

plt.title('Marketing Campaign Funnel')
plt.xlabel('Funnel Stage')
plt.ylabel('Number of Users')
plt.tight_layout()
plt.show()
```

```python
# Marketing Spend and Revenue by Platform

platform_performance = df.groupby('Platform').agg({
    'Marketing_Spend': 'sum',
    'Revenue_Generated': 'sum'
}).sort_values('Revenue_Generated', ascending=False)

print(platform_performance)
```

```python
# Visualize Marketing Spend vs Revenue

platform_performance.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Marketing Spend vs Revenue by Platform')
plt.xlabel('Marketing Platform')
plt.ylabel('Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Platform generating the highest total revenue

best_revenue_platform = platform_performance['Revenue_Generated'].idxmax()

print("Highest Revenue Platform:", best_revenue_platform)
print(
    "Revenue:",
    platform_performance.loc[best_revenue_platform, 'Revenue_Generated']
)
```

```python
# Leads and conversions by marketing platform

conversion_analysis = df.groupby('Platform').agg({
    'Leads_Generated': 'sum',
    'Conversions': 'sum'
}).sort_values('Conversions', ascending=False)

print(conversion_analysis)
```

```python
# Visualize conversions by platform

conversion_analysis.plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Leads and Conversions by Marketing Platform')
plt.xlabel('Marketing Platform')
plt.ylabel('Number of Leads / Conversions')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_conversion_platform = conversion_analysis['Conversions'].idxmax()

print("Highest Conversion Platform:", best_conversion_platform)
print(
    "Total Conversions:",
    conversion_analysis.loc[best_conversion_platform, 'Conversions']
)
```

```python
# Detailed platform performance

final_analysis = df.groupby('Platform').agg({
    'Marketing_Spend': 'sum',
    'Revenue_Generated': 'sum',
    'Conversions': 'sum'
})

# Calculate ROI from total spend and revenue
final_analysis['Calculated_ROI_%'] = (
    (final_analysis['Revenue_Generated'] - final_analysis['Marketing_Spend'])
    / final_analysis['Marketing_Spend']
) * 100

# Calculate conversion rate from leads
leads_by_platform = df.groupby('Platform')['Leads_Generated'].sum()

final_analysis['Conversion_Rate_%'] = (
    final_analysis['Conversions'] / leads_by_platform
) * 100

final_analysis = final_analysis.sort_values(
    'Calculated_ROI_%',
    ascending=False
)

print(final_analysis.round(2))
```

```python
# Final ROI comparison

plt.figure(figsize=(10, 6))

final_analysis['Calculated_ROI_%'].plot(kind='bar')

plt.title('Calculated ROI by Marketing Platform')
plt.xlabel('Marketing Platform')
plt.ylabel('ROI (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
final_analysis.round(2).to_csv(
    '/content/platform_performance_analysis.csv'
)

print("Final analysis saved successfully!")
```

```python
# Final business insights table

report_table = final_analysis[
    ['Marketing_Spend',
     'Revenue_Generated',
     'Conversions',
     'Calculated_ROI_%',
     'Conversion_Rate_%']
].round(2)

print(report_table)
```

```python
# Save the final business insights table

report_table.to_csv(
    '/content/marketing_business_insights.csv'
)

print("Business insights table saved successfully!")
```

```python
# Identify the top 2 channels based on calculated ROI

top_2_channels = final_analysis['Calculated_ROI_%'].head(2)

print("Top 2 Recommended Marketing Channels:")
print(top_2_channels)
```
